```python
import pandas as pd

df = pd.read_csv("data/iris.csv")

df.head()
```

### Loading the Dataset

The Iris dataset was loaded successfully using Pandas.
The `head()` function displays the first five rows of the dataset.
The dataset contains flower measurements and their species.

```python
print("Last five rows:")
display(df.tail())

print("Dataset shape:")
print(df.shape)

print("Column names:")
print(df.columns.tolist())

print("Data types:")
print(df.dtypes)
```

### Dataset Exploration

The `tail()` function displays the last five rows of the dataset.

The `shape` attribute shows the number of rows and columns.
The `columns` attribute displays the column names.
The `dtypes` attribute shows the data type of each column.

```python
print("Missing values per column:")
print(df.isnull().sum())

print("\nNumber of duplicate rows:")
print(df.duplicated().sum())

print("\nUnique species:")
print(df["species"].unique())

print("\nSpecies counts:")
print(df["species"].value_counts())
```

### Data Quality and Species Analysis

This section checks for missing values and duplicate rows.
It also identifies the unique flower species and counts the rows for each species.

```python
print("Mean sepal length:")
print(df["sepal_length"].mean())

print("\nMaximum petal length:")
print(df["petal_length"].max())

print("\nMinimum sepal width:")
print(df["sepal_width"].min())

print("\nMean measurements by species:")
print(df.groupby("species").mean(numeric_only=True))

print("\nStatistical summary:")
print(df.describe())
```

### Statistical Analysis

The mean function calculates the average sepal length.
The max function finds the maximum petal length.
The min function finds the minimum sepal width.
The groupby function calculates the average measurements for each species.
The describe function provides statistical summaries, including count, mean, standard deviation, minimum, and maximum values.

```python
print("Flowers with sepal length greater than 6:")
print(df[df["sepal_length"] > 6])

print("\nFlowers with petal width greater than 1.5:")
print(df[df["petal_width"] > 1.5])

print("\nSetosa flowers:")
print(df[df["species"] == "setosa"])

print("\nSpecies with the highest average petal length:")
print(df.groupby("species")["petal_length"].mean().idxmax())

print("\nDataset sorted by sepal length (descending):")
print(df.sort_values("sepal_length", ascending=False))
```

### Data Filtering and Sorting

The first filter selects flowers with sepal length greater than 6.
The second filter selects flowers with petal width greater than 1.5.
The third filter selects only Setosa flowers.
The groupby operation identifies the species with the highest average petal length.
The sort_values function sorts the dataset by sepal length in descending order.

```python
import os

os.makedirs("output", exist_ok=True)

species_summary = df.groupby("species").mean(numeric_only=True)
species_summary.to_csv("output/species_summary.csv")

filtered_iris = df[df["sepal_length"] > 6]
filtered_iris.to_csv("output/filtered_iris.csv", index=False)

print("Files saved successfully:")
print("output/species_summary.csv")
print("output/filtered_iris.csv")
```

### Exporting Results

The output folder stores the analysis results.
The species_summary.csv file contains the average numerical measurements for each species.
The filtered_iris.csv file contains flowers with sepal length greater than 6.

```python
import matplotlib.pyplot as plt

# Scatter plot
plt.figure(figsize=(8, 5))
plt.scatter(df["sepal_length"], df["petal_length"])
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.title("Sepal Length vs Petal Length")
plt.show()

# Bar chart
df.groupby("species")["petal_length"].mean().plot(kind="bar")
plt.xlabel("Species")
plt.ylabel("Average Petal Length")
plt.title("Average Petal Length by Species")
plt.show()
```

### Data Visualization

The scatter plot shows the relationship between sepal length and petal length.

The bar chart compares the average petal length for each flower species.
